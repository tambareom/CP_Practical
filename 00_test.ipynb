```python
# Student grade A - above 95 marks, B - above 85 marks, C - above 75 marks, D - above 65 marks, E - above 55 marks, F - below 55 marks
marks = float(input("Enter your marks: "))

if marks > 95:
    grade = "A"
elif marks > 85:
    grade = "B"
elif marks > 75:
    grade = "C"
elif marks > 65:
    grade = "D"
elif marks > 55:
    grade = "E"
else:
    grade = "F"

print(f"Your grade is: {grade}")
```
